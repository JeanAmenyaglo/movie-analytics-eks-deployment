```python
# DO NOT CHANGE THIS CELL

import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
```

## Load the Movie Data

```python
# EXPORT
def load_data():
    MOVIE_DATA_PATH = '../movie_dataset.csv'
    # YOUR CODE GOES HERE
      #########################################
    
    # Load the movie dataset from CSV file
    try:
        movies_df = pd.read_csv(MOVIE_DATA_PATH)
        return movies_df
    except FileNotFoundError:
        # Try alternative path if file not found
        try:
            movies_df = pd.read_csv('movie_dataset.csv')
            return movies_df
        except FileNotFoundError:
            print(f"Error: Could not find movie dataset at {MOVIE_DATA_PATH} or in current directory")
            return None
    
    #########################################
```

```python
# DO NOT CHANGE THIS CELL
# Part 2.4 Output.
# this line below should run the data

movie_data = load_data()

# this should show the first 5 lines for the dataframe
movie_data.head()
```

```python
# EXPORT
def filter_top_rated_movies(df):
    # Note to test this cell please


    # YOUR CODE GOES HERE
    #########################################
# Filter the top 15 movies and sort by Rating (descending)
    top_15_movies = df.nlargest(15, 'Rating').sort_values('Rating', ascending=False)
    
    return top_15_movies
    #########################################
```

```python
# DO NOT CHANGE CELL
df_top_rated_movies = filter_top_rated_movies(movie_data)

df_top_rated_movies.head(15)
```

```python
# EXPORT
def plot_bar_chart(df):
    # YOUR CODE GOES BELOW
    #####################################
 # Create a figure with specified size
    plt.figure(figsize=(14, 8))
    
    # Create horizontal bar chart
    bars = plt.barh(df['Title'], df['Rating'], color='steelblue')
    
    # Customize the chart
    plt.xlabel('Rating (out of 10)', fontsize=12)
    plt.ylabel('Movie Title', fontsize=12)
    plt.title('Top Rated Movies', fontsize=16, fontweight='bold')
    
    # Invert y-axis so highest rating appears at top
    plt.gca().invert_yaxis()
    
    # Add rating values on the end of each bar
    for bar, rating in zip(bars, df['Rating']):
        plt.text(bar.get_width() + 0.02,  # Position text slightly to the right of bar
                bar.get_y() + bar.get_height()/2,  # Center vertically
                f'{rating:.1f}',  # Format rating to 1 decimal place
                va='center', ha='left', fontsize=10)
    
    # Add grid for better readability
    plt.grid(axis='x', alpha=0.3, linestyle='--')
    
    # Adjust layout to prevent label cutoff
    plt.tight_layout()
    
    # Show the plot
    plt.show()
    #####################################
```

```python
# DO NOT CHANGE OR REMOVE

plot_bar_chart(df_top_rated_movies)
```

# Fundamental Filters

Using knowledge of [Boolean Indexing](https://pandas.pydata.org/docs/user_guide/10min.html#boolean-indexing) select all movies less than a specific year and sort them by highest rating first, and return the first ten


```python
# year is an integer.
# df is the dataframe.
def filter_movies_by_year(df, year):
    # YOUR CODE BELOW
    ###################################
  # Step 1: Select all movies less than the given year using boolean indexing
    filtered_movies = df[df['Year of release'] < year]
    
    # Step 2: Sort by rating in descending order
    sorted_movies = filtered_movies.sort_values('Rating', ascending=False)
    
    # Step 3: Return only the top 10 movies
    top_10_movies = sorted_movies.head(10)
    
    return top_10_movies
    ###################################
```

```python
# DO NOT CHANGE OR REMOVE

filtered_movies =  filter_movies_by_year(movie_data, 1990)

filtered_movies.head()
```

```python
# Plot the data with a bar chart
# YOUR CODE BELOW
###############################

# Using the plot_bar_chart function we already created in CELL 7
plot_bar_chart(filtered_movies)


###############################
```
